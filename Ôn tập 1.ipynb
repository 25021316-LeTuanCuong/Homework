## Bài 1
Yêu cầu. Cước phí Taxi Thông Minh.
Viết một hàm duy nhất smart_taxi_fare(car_type, start_hour, km, wait_minutes) tính và trả về total (tổng tiền phải trả, 
kiểu số nguyên, đơn vị VND). 
Các quy tắc cần cài đặt: 
1. Giá mở cửa (0–0.8 km): 
- Xe 4 chỗ: 11,000 VND 
- Xe 7 chỗ: 12,000 VND 
2. Giá theo km sau 0.8 km: 
- Xe 4 chỗ: từ 0.8→20 km ơnh 13,000 VND/km; trên 20 km tính 10,500 VND/km 
- Xe 7 chỗ: từ 0.8→20 km ơnh 14,000 VND/km; trên 20 km tính 12,000 VND/km 
3. Phí chờ: 5 phút đầu miễn phí; sau đó, mỗi 5 phút (làm tròn lên theo từng block 5 phút) tính 2,000 VND. 
4. Phụ thu ban đêm: nếu start_hour thuộc [22,23,0,1,2,3,4,5], cộng +20% trên tiền tạm tính (mở cửa + quãng đường + 
chờ). 
5. Điều kiện đặc biệt: 
- Nếu km > 30 và không phải ban đêm, áp dụng giảm 10% trên tổng tiền. 
- Nếu tổng tiền cuối cùng < 50,000 nhưng km > 5, đặt tổng tiền bằng 50,000 VND (mức tối thiểu).

```python
import math
def smart_taxi_fare(car_type, start_hour, km, wait_minutes):
    total = 0
    # Xe 4 chỗ
    if car_type == 4:
        # Xử lý giá xe tính theo quãng đường
        if 0 < km <= 0.8:
            total += 11000
        elif 0.8 < km <= 20:
            total += 11000 + (km - 0.8) * 13000
        elif km > 20:
            total += 11000 + 19.2 * 13000 + (km - 20) * 10500
    
    # Xe 7 chỗ
    elif car_type == 7:
        if 0 < km <= 0.8:
            total += 12000
        elif 0.8 < km <= 20:
            total += 12000 + (km - 0.8) * 14000
        elif km > 20:
            total += 12000 + 19.2 * 14000 + (km - 20) * 12000
    
    # Thời gian chờ
    if wait_minutes > 5:
        total += math.ceil((wait_minutes - 5)/5) * 2000
    # Phụ thu ban đêm 
    if start_hour in [22, 23, 0, 1, 2, 3, 4, 5]:
        total = total * 1.2
    # Điều kiện đặc biệt
    if km > 30 and (start_hour in [22, 23, 0, 1, 2, 3, 4, 5]) == False:
        total = total * 0.9
    if total < 50000 and km > 5:
        total = 50000
    return int(total)

smart_taxi_fare(8, 22, 35, 13)
```

và False nếu ngược lại. 
- Đầu vào: 
- quad: danh sách hoặc bộ gồm đúng 4 đỉnh [(x1,y1), (x2,y2), (x3,y3), (x4,y4)] theo thứ tự (tứ giác có thể lồi hoặc lõm, 
nhưng không tự cắt). 
- p: một điểm (x, y). 
- Đầu ra: giá trị boolean. 
- Gợi ý: Cài đặt phép kiểm tra điểm trong đa giác tổng quát (ví dụ: ray casting), đồng thời tính cả trường hợp điểm nằm 
trên biên là ở trong

```python
def point_in_quadrilateral(quad, p):
    v1 = [p[0] - quad[0][0], p[1] - quad[0][1]]
    v2 = [p[0] - quad[1][0], p[1] - quad[1][1]]
    v3 = [p[0] - quad[2][0], p[1] - quad[2][1]]
    v4 = [p[0] - quad[3][0], p[1] - quad[3][1]]
    vecto1 = [quad[1][0] - quad[0][0], quad[1][1] - quad[0][1]]
    vecto2 = [quad[3][0] - quad[0][0], quad[3][1] - quad[0][1]]
    
    tugiac = abs(vecto1[0]*vecto2[1] - vecto2[0]*vecto1[1])
    tamgiac1 = 1/2 * abs(v1[0]*v2[1] - v2[0]*v1[1])
    tamgiac2 = 1/2 * abs(v2[0]*v3[1] - v3[0]*v2[1])
    tamgiac3 = 1/2 * abs(v3[0]*v4[1] - v4[0]*v3[1])
    tamgiac4 = 1/2 * abs(v1[0]*v4[1] - v4[0]*v1[1])

    if tugiac >= (tamgiac1 + tamgiac2 + tamgiac3 + tamgiac4):
        return True
    else:
        return False
    
point_in_quadrilateral([(0,0), (4,0), (4,3), (0,3)], [2, 1])
```

Yêu cầu: Viết hàm detect_number_type(s) trả về 'int' nếu s biểu diễn một số nguyên, 'float' nếu biểu diễn một số thực dấu 
chấm động, ngược lại trả về 'neither'. 
- Đầu vào: chuỗi s. 
- Đầu ra: một trong các giá trị {'int', 'float', 'neither'}. 
- Quy tắc: 
- Cho phép có dấu + hoặc - ở đầu. 
- Số nguyên phải chỉ gồm các chữ số và không có dấu chấm thập phân (ví dụ: '100'). 
- Số thực phải có dấu chấm và có chữ số ở cả hai phía của dấu chấm (ví dụ: '100.0', '-3.14'). 
- Các chuỗi như '.5', '5.', '1e3' được xem là 'neither' trong bài này.

```python
def detect_number_type(s):
    temp_s = s
    
    if temp_s.startswith('+') or temp_s.startswith('-'):
        temp_s = temp_s[1:]
    
    if not temp_s:
        return 'neither'
    
    if temp_s.isdigit():
        return 'int'
    
    if temp_s.count('.') == 1:
        parts = temp_s.split('.')
        if parts[0].isdigit() and parts[1].isdigit():
            return 'float'
            
    return 'neither'

print(detect_number_type("100"))      
```

- Đầu ra: giá trị boolean.

```python
def is_perfect_square(n):
    if n<0:
        return False
    for i in range(1,n+1):
        if n==i**2:
            return True
    return False
print(is_perfect_square(4))    
    
```

```python
def compute(a,b,op):
    s=['+','-','*','/']
    if op not in s:
        raise ValueError
    if op=='+':
        return a+b
    if op=='-':
        return a-b
    if op=='*':
        return a*b
    if op=='/':
        if b==0:
            raise ValueError
        
        else:
            return a/b
a=float(input())
b=float(input())
op=input()
print(round(compute(a,b,op),2))
    
```

## Bài 43

tạo thành tam giác.

```python
def dientich(a,b,c):
    if a+b<c or a+c<b or b+c<a:
        raise ValueError
    p=a+b+c
    s=(p*(p-a)*(p-b)*(p-c))**0.5
    return s
print(dientich(2,3,4))
```

## Bài 44
Viết hàm trả về tích các số nguyên từ 1 đến n (giai thừa). Với n=0, trả về 1.

```python
def giaithua(n):
    if n == 0:
        return 1
    else:
        s = 1
        for i in range(1, n + 1):
            s = s * i
        return s
print(giaithua(3))
```
