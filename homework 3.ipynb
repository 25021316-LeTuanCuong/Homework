## Bài 9
 Nhập vào số nguyên dương a, kiểm tra xem a có phải là số chính phương hay 
không? Số chính phương là số có căn bậc hai là một số nguyên VD: 1, 4, 9, 16, 25….

```python
a = int(input())
la_so_chinh_phuong = False # Giả sử ban đầu là không phải

for i in range(1, a + 1):
    if i * i == a:
        la_so_chinh_phuong = True
        break # Tìm thấy rồi thì dừng lại luôn cho nhanh

if la_so_chinh_phuong:
    print("Là số chính phương")
else:
    print("Không phải số chính phương")
```

## Bài 10

## Bài 14
Nhập vào một dãy số nguyên, dừng nhập khi người dùng nhập -1. Sau khi nhập xong, in số lớn nhất, số nhỏ nhất trong những số vừa nhập. KHÔNG ĐƯỢC dùng mảng

```python
nums = int(input())
maxnums = nums
minnums = nums
while nums != -1:
    nums = int(input())
    if nums >= maxnums:
        maxnums = nums
    elif nums <= minnums:
        minnums = nums
print(maxnums, minnums)
```

## Bài 15 
Nhập vào số nguyên dương n, đếm xem n có bao nhiêu chữ số. Làm bằng nhiều cách nhất có thể, ko dùng mảng.

## Bài 20
Nhập vào số nguyên dương n, đếm xem n có bao nhiêu chữ số là số nguyên tố.

```python
n = int(input())
dem = 0
num = str(n)
for i in num:
    if i in "0,1":
        continue
    else:
        for k in range(2, int(i)):
            if int(i) % k == 0:
                break
        else:
            dem += 1
print(dem)
```
