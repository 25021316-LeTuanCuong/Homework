## Bài 1
Viết hàm kiểm tra một số có phải số Amstronhg hay không? Hàm trả về kiểu bool, có tham số là một số 
nguyên x

```python
def Amstronhg(a):
    chuoiso=str(a)
    sochuso=len(chuoiso)
    tong=0
    for kytu in chuoiso:
        chuso=int(kytu)
        tong=chuso**sochuso
    return tong==a

so_nguyen = int(input("Nhập một số nguyên bất kỳ: "))

if Amstronhg(so_nguyen):
    print(f"Số {so_nguyen} là số Armstrong.")
else:
    print(f"Số {so_nguyen} không phải là số Armstrong.")
```

## Bài 3
VIết hàm kiểm tra một số có phải số đối xứng không, xử lý trên dữ liệu kiểu số nguyên.

```python
def sodoixung(n):
    n2 = n
    number = 0
    while n2 % 10 != 0:
        number = number * 10 + n2 % 10
        n2 = n2 // 10
    if number == n:
        return True
    else:
        return False

n=int(input())
print(sodoixung(n))
```

## Bài 4
VIết hàm kiểm tra một số có phải số kỳ diệu hay không, xử lý trên dữ liệu kiểu số nguyên

```python
def sokidieu(n):
    s = 0
    for i in range(1, n):
        if n % i == 0:
            s += i
    if s == n:
        return True
    else:
        return False

n=int(input())
print(sokidieu(n))
```

## Bài 5
Viết hàm detect_number_type(s) trả về 'int' nếu s là số nguyên, 'float' nếu là số thực, ngược lại 'neither'.

```python
def detect_number_type(s):
    if type(s)==int:
        return 'int'
    elif type(s)==float:
        return 'float'
    else:
        return 'neither'

print(detect_number_type(5.5))    
    
```

Viết hàm are_collinear(x1,y1,x2,y2,x3,y3) kiểm tra ba điểm có thẳng hàng hay không.

```python
def are_collinear(x1,y1,x2,y2,x3,y3):
    v1=x2-x1
    v2=y2-y1
    u1=x3-x1
    u2=y3-y1
    if v1*u2-v2*u1==0:
        return '3 điểm thẳng hàng'
    else:
        return '3 điểm k thẳng hàng'
print(are_collinear(1,2,3,4,5,6))    

```

## Bài 10
Viết hàm point_in_quadrilateral(quad, p) kiểm tra điểm p có nằm trong hoặc trên biên của tứ giác quad hay 
không. Trong đó: quad là một tứ giác (4 đỉnh – 4 điểm), p là một điểm cần kiểm tra. VD: quad = [(0,0), (4,0), (4,3), 
(0,3)], p = (2,1)

```python
def point_in_quadrilateral(quad, p):
    def tinh_tich_co_huong(a, b, c):
        return (b[0] - a[0]) * (c[1] - a[1]) - (b[1] - a[1]) * (c[0] - a[0])

    so_luong_dinh = len(quad)
    danh_sach_dau = []

    for i in range(so_luong_dinh):
        dinh_dau = quad[i]
        dinh_cuoi = quad[(i + 1) % so_luong_dinh]
        tich_so = tinh_tich_co_huong(dinh_dau, dinh_cuoi, p)
        if tich_so != 0:
            danh_sach_dau.append(tich_so > 0)

    if not danh_sach_dau:
        return True
        
    return all(dau == danh_sach_dau[0] for dau in danh_sach_dau)


quad = [(0, 0), (4, 0), (4, 3), (0, 3)]
p = [2, 1]
print(point_in_quadrilateral(quad, p))
```

## Bài 13
Viết hàm tính góc ∠ABC (độ) từ tọa độ A(x1,y1), B(x2,y2), C(x3,y3).

```python
def goc(x1, y1, x2, y2, x3, y3):
    import math
    v1 = [x1 - x2, y1 - y2]
    v2 = [x3 - x2, y3 - y2]
    cos = abs((v1[0]*v2[0] + v1[1]*v2[1])/((v1[0]**2 + v1[1]**2)**0.5 * (v2[0]**2 + v2[1]**2)**0.5))
    return math.acos(cos)*180/math.pi

print(goc(1,2,2,4,5,5))
```

## Bài 14
Viết hàm trả về giá trị lớn nhất và nhỏ nhất của ba số a,b,c dưới dạng list [max, min].

```python
def timmaxmin(a,b,c):
    max=a
    if b>max:
        max=b
    if c>max:  
        max=c
    min=a
    if b<min:
        min=b
    if c<min:
        min=c
    return(max,min)
print(timmaxmin(2,3,4))
```

## Bài 15
Viết hàm tính hệ số nhị thức C(n, k).

```python
def giaithua(n):
    s=1
    for i in range(1,n+1):
        s=s*i
    return s
def nhithuc(n,k):
    return giaithua(n)/(giaithua(k)*giaithua(n-k))
print(nhithuc(9,7))  
```

Nếu n ≤ 0 trả về 0.
